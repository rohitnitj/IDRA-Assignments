```python
import csv
import os

FILE_NAME = 'library_data.csv'

class Book:
    """Class representing an individual Book."""

    def __init__(self, book_id, title, author, is_borrowed=False):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.is_borrowed = is_borrowed

    def __str__(self):
        status = "Borrowed" if self.is_borrowed else "Available"
        return f"[{self.book_id}] '{self.title}' by {self.author} - {status}"

    def to_dictionary(self):
        return {
            'ID': self.book_id,
            'Title': self.title,
            'Author': self.author,
            'Is_Borrowed': str(self.is_borrowed)
        }


class Library:
    """Class managing a collection of Book objects."""

    def __init__(self):
        self.books = {}
        self.load_books()

    def load_books(self):
        """Read data from CSV and populate the books dictionary."""
        if not os.path.exists(FILE_NAME):
            self.save_books()
            return

        try:
            with open(FILE_NAME, mode='r') as file:
                reader = csv.DictReader(file)
                for row in reader:
                    is_borrowed = row['Is_Borrowed'] == 'True'
                    self.books[row['ID']] = Book(row['ID'], row['Title'], row['Author'], is_borrowed)
        except Exception as e:
            print(f"Error loading database: {e}")

    def save_books(self):
        """Save all Book objects to the CSV file."""
        try:
            with open(FILE_NAME, mode='w', newline='') as file:
                fieldnames = ['ID', 'Title', 'Author', 'Is_Borrowed']
                writer = csv.DictWriter(file, fieldnames=fieldnames)
                writer.writeheader()
                for book in self.books.values():
                    writer.writerow(book.to_dictionary())
        except Exception as e:
            print(f"Error saving database: {e}")

    def add_book(self, title, author):
        """Add a new book to the library."""
        new_id = f"B{len(self.books) + 101}"
        new_book = Book(new_id, title, author)
        self.books[new_id] = new_book
        self.save_books()
        return new_id

    def view_all_books(self):
        """Display all books."""
        if not self.books:
            print("The library is currently empty.")
            return

        print("\n--- Library Catalog ---")
        for book in self.books.values():
            print(book)

    def display_unique_authors(self):
        """Use a set to find and display unique authors in the library."""
        authors = {book.author for book in self.books.values()}

        if not authors:
            print("No authors found.")
            return

        print("\n--- Featured Authors ---")
        for author in sorted(authors):
            print(f"- {author}")

    def borrow_book(self, book_id):
        """Mark a book as borrowed. Returns a tuple (success_boolean, message)."""
        if book_id not in self.books:
            return (False, "Book ID not found.")

        book = self.books[book_id]
        if book.is_borrowed:
            return (False, f"'{book.title}' is already borrowed.")

        book.is_borrowed = True
        self.save_books()
        return (True, f"You have successfully borrowed '{book.title}'.")

    def return_book(self, book_id):
        """Mark a book as returned. Returns a tuple (success_boolean, message)."""
        if book_id not in self.books:
            return (False, "Book ID not found.")

        book = self.books[book_id]
        if not book.is_borrowed:
            return (False, f"'{book.title}' was not borrowed.")

        book.is_borrowed = False
        self.save_books()
        return (True, f"You have successfully returned '{book.title}'.")


def main():
    my_library = Library()

    while True:
        print("\n" + "="*35)
        print("   CITY LIBRARY MANAGEMENT SYSTEM")
        print("="*35)
        print("1. Add a New Book")
        print("2. View All Books")
        print("3. View Unique Authors")
        print("4. Borrow a Book")
        print("5. Return a Book")
        print("6. Exit")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == '1':
            title = input("Enter book title: ").strip().title()
            author = input("Enter author name: ").strip().title()
            if title and author:
                new_id = my_library.add_book(title, author)
                print(f"Success: '{title}' added with ID: {new_id}")
            else:
                print("Error: Title and Author cannot be empty.")

        elif choice == '2':
            my_library.view_all_books()

        elif choice == '3':
            my_library.display_unique_authors()

        elif choice == '4':
            book_id = input("Enter the Book ID to borrow (e.g., B101): ").strip().upper()
            success, message = my_library.borrow_book(book_id)
            if success:
                print(f"Success: {message}")
            else:
                print(f"Error: {message}")

        elif choice == '5':
            book_id = input("Enter the Book ID to return (e.g., B101): ").strip().upper()
            success, message = my_library.return_book(book_id)
            if success:
                print(f"Success: {message}")
            else:
                print(f"Error: {message}")

        elif choice == '6':
            print("Saving data and exiting... Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
