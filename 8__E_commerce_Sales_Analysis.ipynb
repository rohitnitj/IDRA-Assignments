#1: Importing Libraries and Loading Data

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

# Inspect the first 5 rows
print("--- First 5 Rows ---")
display(df.head())

# Get basic information about the dataset (columns, data types, non-null counts)
print("\n--- Dataset Info ---")
df.info()
```

#2: General Sales Statistics

```python
# Calculate basic statistics
total_sales = df['Total_Sales'].sum()
average_sales = df['Total_Sales'].mean()
max_sales = df['Total_Sales'].max()
min_sales = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
total_orders = df['Order_ID'].nunique()

# Print the statistics
print("--- Overall Sales Statistics ---")
print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Sales per Order: ₹{average_sales:,.2f}")
print(f"Highest Single Order Sales: ₹{max_sales:,.2f}")
print(f"Lowest Single Order Sales: ₹{min_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity} items")
print(f"Total Number of Orders: {total_orders}")
```

#3: Grouping and Aggregation (Category & City)

```python
# 1. Total Sales by Category
category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False).reset_index()
print("--- Total Sales by Category ---")
display(category_sales)

# 2. Total Sales by City (Top 5 Cities)
city_sales = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False).head(5).reset_index()
print("\n--- Top 5 Cities by Sales ---")
display(city_sales)
```

#4: Grouping and Aggregation (Products & Payments)

```python
# 1. Top 5 Best-Selling Products
top_products = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False).head(5).reset_index()
print("--- Top 5 Best-Selling Products ---")
display(top_products)

# 2. Payment Method Popularity (Count of orders)
payment_methods = df['Payment_Method'].value_counts().reset_index()
payment_methods.columns = ['Payment_Method', 'Order_Count']
print("\n--- Payment Method Distribution ---")
display(payment_methods)
```

#5: Filtering and Sorting Data

```python
# Filter: Identify orders where Total_Sales is greater than ₹10,000
high_value_orders = df[df['Total_Sales'] > 10000]
print(f"--- High Value Orders (> ₹10,000) : {len(high_value_orders)} orders ---")
display(high_value_orders[['Order_ID', 'Customer', 'Category', 'Total_Sales']])

# Sort: Top 5 Highest Individual Orders in the dataset
top_5_orders = df.sort_values(by='Total_Sales', ascending=False).head(5)
print("\n--- Top 5 Highest Individual Orders ---")
display(top_5_orders[['Order_ID', 'Customer', 'Product', 'Quantity', 'Total_Sales']])
```
