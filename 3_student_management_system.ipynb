```python
students = []

def add_student():
    """Adds a new student to the system."""
    print("\n--- Add New Student ---")
    try:
        std_id = input("Enter Student ID: ")

        if any(student['ID'] == std_id for student in students):
            print("Error: Student ID already exists!")
            return

        name = input("Enter Name: ")
        age = int(input("Enter Age: "))
        course = input("Enter Course: ")
        marks = float(input("Enter Marks: "))

        student_record = {
            "ID": std_id,
            "Name": name,
            "Age": age,
            "Course": course,
            "Marks": marks
        }

        students.append(student_record)
        print(f"Success: Student '{name}' added successfully!")

    except ValueError:
        print("Error: Invalid input. Age and Marks must be numeric values.")

def view_students():
    """Displays all student records."""
    print("\n--- All Student Records ---")
    if not students:
        print("No student records found.")
        return

    for s in students:
        print(f"ID: {s['ID']} | Name: {s['Name']} | Age: {s['Age']} | Course: {s['Course']} | Marks: {s['Marks']}")
    print("-" * 27)

def search_student():
    """Searches for a student by ID or Name."""
    print("\n--- Search Student ---")
    query = input("Enter Student ID or Name to search: ").strip().lower()
    found = False

    for s in students:
        if s['ID'].lower() == query or s['Name'].lower() == query:
            print("\nStudent Found:")
            print(f"ID: {s['ID']} | Name: {s['Name']} | Age: {s['Age']} | Course: {s['Course']} | Marks: {s['Marks']}")
            found = True

    if not found:
        print("Student not found.")

def update_student():
    """Updates an existing student's details based on ID."""
    print("\n--- Update Student ---")
    std_id = input("Enter Student ID to update: ")

    for s in students:
        if s['ID'] == std_id:
            print("Leave the field blank and press Enter to keep the current value.")

            name = input(f"Enter new Name ({s['Name']}): ")
            age_str = input(f"Enter new Age ({s['Age']}): ")
            course = input(f"Enter new Course ({s['Course']}): ")
            marks_str = input(f"Enter new Marks ({s['Marks']}): ")

            try:
                if name: s['Name'] = name
                if age_str: s['Age'] = int(age_str)
                if course: s['Course'] = course
                if marks_str: s['Marks'] = float(marks_str)

                print("Success: Student record updated successfully!")
            except ValueError:
                print("Error: Invalid input for Age or Marks. Update aborted.")
            return

    print("Error: Student ID not found.")

def delete_student():
    """Deletes a student record based on ID."""
    print("\n--- Delete Student ---")
    std_id = input("Enter Student ID to delete: ")

    for i, s in enumerate(students):
        if s['ID'] == std_id:
            name = s['Name']
            del students[i]
            print(f"Success: Student '{name}' deleted successfully!")
            return

    print("Error: Student ID not found.")

def main():
    """Main menu loop."""
    while True:
        print("\n" + "="*30)
        print(" STUDENT MANAGEMENT SYSTEM")
        print("="*30)
        print("1. Add New Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Exit")

        choice = input("Enter your choice (1-6): ")

        if choice == '1':
            add_student()
        elif choice == '2':
            view_students()
        elif choice == '3':
            search_student()
        elif choice == '4':
            update_student()
        elif choice == '5':
            delete_student()
        elif choice == '6':
            print("Exiting Student Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
